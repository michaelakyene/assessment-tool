## Concept Explanation (Simple → Deep)

Data science is the practice of using data to answer questions and guide decisions.

### 2) Practical View
A data scientist collects and prepares data, explores it, builds models, and communicates insights to stakeholders.

### 3) Professional View
In real organizations, data science combines statistics, software engineering, domain knowledge, and communication. Success is measured by business impact, not only model accuracy.

```
# Core imports for this notebook
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

np.random.seed(7)
sns.set_theme(style='whitegrid')
```

## Scenario Exercise: Online Learning Platform
You are a junior data scientist at an online learning company.
Your manager asks:

**'Can we identify patterns that relate to course completion so the student success team can intervene early?'**

You will start with a simulated dataset representing learner behavior.

```
# Create a realistic simulated dataset
n = 300
study_hours = np.random.normal(loc=6, scale=2, size=n).clip(0, 14)
attendance = np.random.normal(loc=80, scale=12, size=n).clip(30, 100)
assignments_submitted = np.random.poisson(lam=7, size=n).clip(0, 10)
discussion_posts = np.random.poisson(lam=4, size=n)

score = (
    35
    + 4.2 * study_hours
    + 0.35 * attendance
    + 2.8 * assignments_submitted
    + 0.9 * discussion_posts
    + np.random.normal(0, 7, size=n)
).clip(0, 100)

completed = (score >= 70).astype(int)

df = pd.DataFrame({
    'student_id': np.arange(1, n + 1),
    'study_hours_per_week': study_hours.round(2),
    'attendance_rate': attendance.round(1),
    'assignments_submitted': assignments_submitted,
    'discussion_posts': discussion_posts,
    'final_score': score.round(1),
    'completed_course': completed
})

df.head()
```

## First Data Inspection
A data scientist should always inspect shape, types, summary stats, and quality before deep analysis.

```
print('Shape:', df.shape)
display(df.info())
display(df.describe(include='all').T)
```

## Data Cleaning Task

```
# Inject a few missing values for practice
df.loc[df.sample(frac=0.04, random_state=12).index, 'study_hours_per_week'] = np.nan
df.loc[df.sample(frac=0.03, random_state=21).index, 'attendance_rate'] = np.nan

print('Missing values before cleaning:')
display(df.isna().sum())

# Clean with robust defaults
df['study_hours_per_week'] = df['study_hours_per_week'].fillna(df['study_hours_per_week'].median())
df['attendance_rate'] = df['attendance_rate'].fillna(df['attendance_rate'].mean())

print('Missing values after cleaning:')
display(df.isna().sum())
```

## Visualization Tasks

```
# 1) Distribution of final scores
plt.figure(figsize=(9, 4))
sns.histplot(df['final_score'], kde=True, bins=25, color='teal')
plt.title('Distribution of Final Scores')
plt.xlabel('Final Score')
plt.ylabel('Number of Students')
plt.show()
```

```
# 2) Completion rate
completion_rate = df['completed_course'].mean() * 100
print(f'Completion rate: {completion_rate:.1f}%')

plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='completed_course', palette='Set2')
plt.xticks([0, 1], ['Not Completed', 'Completed'])
plt.title('Course Completion Counts')
plt.xlabel('Outcome')
plt.ylabel('Count')
plt.show()
```

```
# 3) Relationship between effort and performance
plt.figure(figsize=(8, 5))
sns.scatterplot(
    data=df,
    x='study_hours_per_week',
    y='final_score',
    hue='completed_course',
    palette={0: 'tomato', 1: 'seagreen'},
    alpha=0.75
)
plt.title('Study Hours vs Final Score')
plt.xlabel('Study Hours per Week')
plt.ylabel('Final Score')
plt.legend(title='Completed', labels=['No', 'Yes'])
plt.show()
```

## Applied Insight Example
Use grouped analysis to find practical intervention points.

```
grouped = df.groupby('completed_course')[['study_hours_per_week', 'attendance_rate', 'assignments_submitted', 'final_score']].mean().round(2)
grouped.index = ['Not Completed', 'Completed']
grouped
```
