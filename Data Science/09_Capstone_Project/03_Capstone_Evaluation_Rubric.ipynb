# 03 Capstone Evaluation Rubric

**Module:** 09_Capstone_Project

This capstone module transforms your learning into a portfolio-ready, real-world project.

## Learning Objectives
- Define a business-relevant data science problem
- Plan and execute an end-to-end project workflow
- Communicate impact, limitations, and next actions
- Evaluate quality with a professional rubric

## Evaluation Criteria (100 points)
- Problem framing and business relevance: **15**
- Data quality handling and feature design: **20**
- EDA depth and insight quality: **15**
- Modeling/statistical rigor: **20**
- Communication and visualization clarity: **15**
- Reproducibility and code quality: **10**
- Reflection, limitations, and next steps: **5**

## Performance Bands
- **Outstanding (90–100):** clear business impact, robust methods, and professional communication
- **Strong (75–89):** sound workflow with minor gaps
- **Developing (60–74):** partial execution, weak linkage between analysis and decisions
- **Needs Work (<60):** major reproducibility or reasoning gaps

## Challenge Task
Self-assess your capstone using this rubric, then list top 3 upgrades to move one performance band higher.

## Summary
Use this rubric as both grading criteria and a quality checklist before publishing your portfolio project.
