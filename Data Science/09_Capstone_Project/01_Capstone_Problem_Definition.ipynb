# 01 Capstone Problem Definition

**Module:** 09_Capstone_Project

This capstone module transforms your learning into a portfolio-ready, real-world project.

## Learning Objectives
- Define a business-relevant data science problem
- Plan and execute an end-to-end project workflow
- Communicate impact, limitations, and next actions
- Evaluate quality with a professional rubric

## Real-World Problem Definition
**Problem:** A telecom company has rising customer churn and wants to proactively retain at-risk users.

### Business Goal
Reduce churn by 10% over two quarters while preserving marketing efficiency.

### Data Science Goal
Build a churn risk segmentation solution and identify key behavioral drivers of churn.

## Dataset Suggestions
- IBM Telco Customer Churn dataset
- Bank Marketing dataset
- E-commerce subscription retention data (public/open)
- Simulated SaaS churn dataset in `Data Science/datasets/`

## Practice Exercises
1. Rewrite the problem statement for healthcare or education.
2. Define at least 3 measurable success metrics.
3. Identify bias and fairness risks in the available features.

## Challenge Task
Draft a one-page project charter with scope, constraints, and expected impact.

## Summary
A strong capstone starts with sharp problem framing, realistic constraints, and measurable outcomes.
