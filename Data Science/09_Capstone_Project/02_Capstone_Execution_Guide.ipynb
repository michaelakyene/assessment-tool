## Step-by-Step Project Guidance
1. Frame the problem and define stakeholders.
2. Acquire and document data sources.
3. Perform cleaning and feature engineering.
4. Conduct exploratory analysis and identify hypotheses.
5. Build baseline and improved models (if predictive).
6. Evaluate with technical + business metrics.
7. Communicate recommendations with visuals.
8. Define deployment and monitoring plan.
