```
# Core imports for this course
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, mean_squared_error

np.random.seed(42)
sns.set_theme(style='whitegrid')
```

## Real-World Dataset / Simulated Dataset
Use one of the following in this notebook:
- Local file in `Data Science/datasets/`
- Public CSV (e.g., government open data)
- Simulated dataset generated with NumPy/Pandas

```
# Example simulated dataset
rows = 200
df = pd.DataFrame({
    'customer_id': np.arange(1, rows + 1),
    'monthly_spend': np.random.normal(120, 35, rows).round(2),
    'sessions': np.random.poisson(10, rows),
    'churn': np.random.binomial(1, 0.25, rows)
})
df.head()
```

## Data Cleaning Task

```
# Insert missing values for practice
df.loc[df.sample(frac=0.05, random_state=42).index, 'monthly_spend'] = np.nan

# Cleaning example
df['monthly_spend'] = df['monthly_spend'].fillna(df['monthly_spend'].median())
df.isna().sum()
```

## Visualization Task

```
plt.figure(figsize=(8, 4))
sns.histplot(df['monthly_spend'], kde=True, bins=25)
plt.title('Distribution of Monthly Spend')
plt.xlabel('Monthly Spend')
plt.ylabel('Count')
plt.show()
```
